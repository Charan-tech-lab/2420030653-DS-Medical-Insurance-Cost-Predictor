```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
p=0.3
data = np.random.binomial(1,p,1000)
sns.histplot(data, stat="density", discrete=True)
plt.title(f'Binomial Distribution (n=1, p={p})')
plt.xlabel('Outcome(0=failure, 1=success)')
plt.ylabel('Density')
plt.xticks([0, 1])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
Iam=4
data=np.random.poisson(Iam,1000)
sns.histplot(data,kde=False,stat="density",bins=range(min(data),max(data)+1))
plt.title(f'Poisson Distribution (lambda={Iam})')
plt.xlabel("Number of events")
plt.ylabel("Density")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
mu=0
sigma=1
data = np.random.normal(mu, sigma, 1000)
sns.histplot(data,kde=True,stat="density")
plt.title(f'Normal Distribution (mean={mu}, std={sigma})')
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
low=0
high=1
data = np.random.uniform(low, high, 1000)
sns.histplot(data,kde=True,stat="density")
plt.title(f'Uniform Distribution ({low}, {high})')
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
Iam=1.5
data=np.random.exponential(Iam,1000)
sns.histplot(data,kde=True,stat="density")
plt.title(f'Exponential Distribution (lambda={Iam})')
plt.xlabel("Time")
plt.ylabel("Density")
plt.show()
```

```python
import pandas as pd
data = pd.DataFrame({
    'Student': ['A', 'B', 'C', 'D', 'E','F'],
    'Department':['MCA','MCA','BTech','BTech','MBA','MBA']})
sample=data.groupby('Department').sample(n=2, random_state=1)
print("Stratified Sample:")
print(sample)
```

```python
import pandas as pd
data=pd.DataFrame({
    'Student': ['A', 'B', 'C', 'D', 'E','F'],
    'Class' : ['C1', 'C1', 'C2', 'C2', 'C3', 'C3']
})
cluster_sample=data[data['Class'] == 'C2']
print("Cluster Sample:")
print(cluster_sample)
```

```python
import numpy as np
data=np.arange(1,101)
sample_without_replacement=np.random.choice(data, size=10, replace=False)
sample_with_replacement=np.random.choice(data, size=10, replace=True)
print("Sample without replacement:",sample_without_replacement)
print("Sample with replacement:",sample_with_replacement)
```

```python
import pandas as pd
df=pd.DataFrame({
    'Study_Hours': [2, 4, 6, 8, 10],
    'Marks':[20, 40, 60, 80, 100]
})
correlation=df['Study_Hours'].corr(df['Marks'])
print("Correlation Coefficient:", correlation)
```

```python
import pandas as pd
df=pd.DataFrame({
        'Study_Hours': [2, 4, 6, 8, 10],
    'Marks':[20, 40, 60, 80, 100]
})
covariance=df['Study_Hours'].cov(df['Marks'])
print("Covariance:", covariance)
```
