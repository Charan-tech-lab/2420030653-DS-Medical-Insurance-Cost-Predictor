```python
import matplotlib.pyplot as plt

y1=[]
y2=[]
x=range(-100,100,10)
for i in x:
    y1.append(i ** 2)
for i in x:
    y2.append(-i**2)
plt.plot(x,y1,color='green')
plt.plot(x,y2,color='Maroon')
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("simple graph")
plt.axhline(color='red')
plt.axvline(color='green')
plt.grid(True)
plt.show()

```
